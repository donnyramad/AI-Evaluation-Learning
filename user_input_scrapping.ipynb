```python
import requests
from bs4 import BeautifulSoup

def scrape_url():
    # 1. Meminta input dari user
    url = input("Masukkan URL website (contoh: https://www.wikipedia.org): ")

    # 2. Persiapan request
    headers = {'User-Agent': 'Mozilla/5.0'}

    try:
        response = requests.get(url, headers=headers)
        response.raise_for_status() # Cek apakah link bisa diakses

        soup = BeautifulSoup(response.text, 'html.parser')

        # 3. Mengambil Judul
        title = soup.find('h1')
        print(f"\n--- Judul: {title.text.strip() if title else 'Judul tidak ditemukan'} ---")

        # 4. Mengambil teks dari semua paragraf
        print("\n--- Ringkasan Konten (3 Paragraf Pertama) ---")
        paragraphs = soup.find_all('p')
        count = 0
        for p in paragraphs:
            text = p.text.strip()
            if text:
                print(f"- {text[:150]}...")
                count += 1
            if count == 3:
                break

    except Exception as e:
        print(f"Terjadi kesalahan: {e}")

# Jalankan fungsinya
scrape_url()
```
