```python
import requests
from bs4 import BeautifulSoup

# Menambahkan 'User-Agent' agar dianggap sebagai browser manusia oleh server
headers = {'User-Agent': 'Mozilla/5.0'}
url = 'https://en.wikipedia.org/wiki/Artificial_intelligence'

response = requests.get(url, headers=headers)
soup = BeautifulSoup(response.text, 'html.parser')

# Menggunakan metode yang lebih aman dengan pengecekan
h1_tag = soup.find('h1')

if h1_tag:
    judul = h1_tag.text
    print(f"Judul Halaman: {judul}")
else:
    print("Judul tidak ditemukan!")

# Mengambil paragraf dengan pengecekan
paragraf_list = soup.find_all('p')

print("\n--- Isi 3 Paragraf Pertama ---")
count = 0
for p in paragraf_list:
    if p.text.strip():  # Hanya ambil paragraf yang ada isinya
        print(f"{count+1}. {p.text[:100]}...")
        count += 1
    if count == 3:
        break
```
